```python
import pandas as pd
```

```python
df = pd.read_csv(r'loan_approval.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop(columns = ['name', 'city', 'points'], axis = 1, inplace = True)
```

```python
df.head()
```

```python
df['loan_approved'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['loan_approved'] = le.fit_transform(df['loan_approved'])
```

```python
df.head()
```

```python
x = df.drop(columns = ['loan_approved'])
y = df['loan_approved']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaling = StandardScaler()
```

```python
x_train = scaling.fit_transform(x_train)
x_test = scaling.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
```

```python
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
models = {
    'Logistic Regression' : LogisticRegression(),
    'Decision Tree' : DecisionTreeClassifier(),
    'Random Forest' : RandomForestClassifier(n_estimators = 100),
    'AdaBoost' : AdaBoostClassifier(n_estimators = 100),
    'GradientBoosting' : GradientBoostingClassifier(),
    'KNeighbors' : KNeighborsClassifier(n_neighbors = 5),
    'SVC' : SVC(),
    'XGBoost' : XGBClassifier(n_estimators = 10)
}
```

```python
results = []

for name, model in models.items():
    model.fit(x_train, y_train)

    y_pred = model.predict(x_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    results.append({
        'Model Name' : name,
        'Accuracy' : accuracy,
        'Precision' : precision,
        'Recall' : recall,
        'F1 Score' : f1
    })

    print('-' * 45)
    print('Model', name)
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    print('Accuracy Score', accuracy)

summary_df = pd.DataFrame(results).sort_values(by = 'Accuracy', ascending = False)
summary_df.index = range(1, len(summary_df) + 1)

print('=' * 45)
print('Model Comparsion Summary')
print('=' * 45)
summary_df
```
